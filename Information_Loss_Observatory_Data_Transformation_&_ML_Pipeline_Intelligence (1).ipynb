```python
# ============================================================
# INFORMATION LOSS OBSERVATORY
# Data Transformation & ML Pipeline Intelligence
# COMPLETE ONE-CELL COLAB PROJECT
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import entropy, ks_2samp

from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from IPython.display import display

np.random.seed(42)

# ============================================================
# 1. PROJECT DIRECTORIES
# ============================================================

BASE_DIR = Path("/content/information_loss_observatory")

DATA_DIR = BASE_DIR / "data"
REPORT_DIR = BASE_DIR / "reports"
FIGURE_DIR = REPORT_DIR / "figures"

for folder in [BASE_DIR, DATA_DIR, REPORT_DIR, FIGURE_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("=" * 100)
print("INFORMATION LOSS OBSERVATORY")
print("=" * 100)

# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def calculate_entropy(series, bins=20):

    values = series.dropna()

    if len(values) == 0:
        return 0.0

    if pd.api.types.is_numeric_dtype(values):

        try:
            values = pd.cut(
                values,
                bins=bins,
                duplicates="drop"
            )
        except Exception:
            pass

    probabilities = values.value_counts(
        normalize=True
    )

    if len(probabilities) == 0:
        return 0.0

    return float(
        entropy(probabilities)
    )


def profile_dataset(df, stage):

    numeric_columns = df.select_dtypes(
        include=np.number
    ).columns

    total_cells = (
        df.shape[0] *
        df.shape[1]
    )

    missing_cells = int(
        df.isna().sum().sum()
    )

    entropy_values = []

    for column in numeric_columns:

        entropy_values.append(
            calculate_entropy(
                df[column]
            )
        )

    mean_entropy = (
        np.mean(entropy_values)
        if entropy_values
        else 0
    )

    return {

        "Stage": stage,

        "Rows":
            int(df.shape[0]),

        "Columns":
            int(df.shape[1]),

        "Total_Cells":
            int(total_cells),

        "Missing_Cells":
            missing_cells,

        "Missing_Rate_%":
            round(
                missing_cells /
                max(total_cells, 1) *
                100,
                3
            ),

        "Duplicate_Rows":
            int(
                df.duplicated().sum()
            ),

        "Unique_Rows":
            int(
                df.drop_duplicates().shape[0]
            ),

        "Numeric_Features":
            len(numeric_columns),

        "Mean_Entropy":
            round(
                float(mean_entropy),
                5
            )
    }


def save_plot(filename):

    plt.tight_layout()

    plt.savefig(
        FIGURE_DIR / filename,
        dpi=180,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# 3. GENERATE RAW DATASET
# ============================================================

N = 5000

raw_df = pd.DataFrame({

    "customer_id":
        np.arange(1, N + 1),

    "age":
        np.random.normal(
            38,
            12,
            N
        ).clip(
            18,
            80
        ),

    "income":
        np.random.lognormal(
            10.7,
            0.45,
            N
        ),

    "transactions":
        np.random.poisson(
            8,
            N
        ),

    "session_duration":
        np.random.gamma(
            4,
            7,
            N
        ),

    "purchase_frequency":
        np.random.poisson(
            3,
            N
        ),

    "support_tickets":
        np.random.poisson(
            2,
            N
        ),

    "satisfaction_score":
        np.random.normal(
            7,
            1.5,
            N
        ).clip(
            1,
            10
        ),

    "discount_usage":
        np.random.beta(
            2,
            5,
            N
        ),

    "website_visits":
        np.random.poisson(
            12,
            N
        ),

    "mobile_usage":
        np.random.beta(
            5,
            2,
            N
        ),

    "account_age_days":
        np.random.randint(
            30,
            2500,
            N
        ),

    "region":
        np.random.choice(
            [
                "North",
                "South",
                "East",
                "West"
            ],
            N,
            p=[
                0.25,
                0.30,
                0.20,
                0.25
            ]
        ),

    "membership":
        np.random.choice(
            [
                "Basic",
                "Silver",
                "Gold",
                "Platinum"
            ],
            N,
            p=[
                0.45,
                0.30,
                0.18,
                0.07
            ]
        )
})


# ============================================================
# 4. CREATE SYNTHETIC TARGET
# ============================================================

risk_score = (

    0.025 *
    raw_df["age"]

    - 0.000006 *
    raw_df["income"]

    - 0.12 *
    raw_df["transactions"]

    - 0.08 *
    raw_df["purchase_frequency"]

    + 0.15 *
    raw_df["support_tickets"]

    - 0.25 *
    raw_df["satisfaction_score"]

    - 0.03 *
    raw_df["website_visits"]

    + 0.8 *
    raw_df["discount_usage"]

    + np.random.normal(
        0,
        1,
        N
    )
)


probability = (
    1 /
    (
        1 +
        np.exp(
            -np.clip(
                risk_score,
                -20,
                20
            )
        )
    )
)


raw_df["churn"] = (
    np.random.random(N)
    < probability
).astype(int)


# ============================================================
# 5. INTRODUCE DATA QUALITY PROBLEMS
# ============================================================

missing_columns = [

    "income",

    "satisfaction_score",

    "session_duration",

    "discount_usage"

]


for column in missing_columns:

    mask = (
        np.random.random(N)
        < 0.05
    )

    raw_df.loc[
        mask,
        column
    ] = np.nan


# Add duplicate records

duplicates = raw_df.sample(
    100,
    random_state=42
)


raw_df = pd.concat(
    [
        raw_df,
        duplicates
    ],
    ignore_index=True
)


raw_df.to_csv(
    DATA_DIR /
    "raw_dataset.csv",
    index=False
)


# ============================================================
# 6. RAW DATASET DISPLAY
# ============================================================

print("\nRAW DATASET")
print("-" * 100)

print(
    "Shape:",
    raw_df.shape
)

display(
    raw_df.head(10)
)


# ============================================================
# 7. STAGE 1 - RAW DATA
# ============================================================

clean_df = raw_df.copy()


# ============================================================
# 8. STAGE 2 - DUPLICATE REMOVAL
# ============================================================

clean_df = (
    clean_df
    .drop_duplicates(
        subset="customer_id"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 9. STAGE 3 - MISSING VALUE IMPUTATION
# ============================================================

imputed_df = clean_df.copy()


numeric_columns = (
    imputed_df
    .select_dtypes(
        include=np.number
    )
    .columns
    .tolist()
)


categorical_columns = (
    imputed_df
    .select_dtypes(
        include="object"
    )
    .columns
    .tolist()
)


for column in numeric_columns:

    if imputed_df[column].isna().any():

        imputed_df[column] = (
            imputed_df[column]
            .fillna(
                imputed_df[column]
                .median()
            )
        )


for column in categorical_columns:

    if imputed_df[column].isna().any():

        imputed_df[column] = (
            imputed_df[column]
            .fillna(
                imputed_df[column]
                .mode()
                .iloc[0]
            )
        )


# ============================================================
# 10. STAGE 4 - OUTLIER FILTERING
# ============================================================

outlier_df = imputed_df.copy()


outlier_mask = pd.Series(
    True,
    index=outlier_df.index
)


outlier_columns = [

    "income",

    "session_duration",

    "support_tickets"

]


for column in outlier_columns:

    q01 = (
        outlier_df[column]
        .quantile(0.01)
    )

    q99 = (
        outlier_df[column]
        .quantile(0.99)
    )

    outlier_mask &= (
        outlier_df[column]
        .between(
            q01,
            q99
        )
    )


outlier_df = (
    outlier_df[
        outlier_mask
    ]
    .reset_index(
        drop=True
    )
)


# ============================================================
# 11. STAGE 5 - FEATURE ENGINEERING
# ============================================================

feature_df = outlier_df.copy()


feature_df[
    "income_per_transaction"
] = (

    feature_df["income"] /
    (
        feature_df["transactions"]
        + 1
    )

)


feature_df[
    "engagement_score"
] = (

    feature_df["website_visits"]

    + feature_df["transactions"]

    + feature_df["purchase_frequency"]

) / 3


feature_df[
    "support_intensity"
] = (

    feature_df["support_tickets"]

    /

    (
        feature_df["account_age_days"]
        / 365
        + 1
    )

)


feature_df[
    "customer_value_signal"
] = (

    feature_df["income"]

    *

    (
        feature_df["purchase_frequency"]
        + 1
    )

)


# ============================================================
# 12. STAGE 6 - FEATURE SELECTION
# ============================================================

selected_features = [

    "age",

    "income",

    "transactions",

    "purchase_frequency",

    "support_tickets",

    "satisfaction_score",

    "website_visits",

    "engagement_score",

    "support_intensity",

    "customer_value_signal"

]


selected_df = (
    feature_df[
        selected_features +
        ["churn"]
    ]
    .copy()
)


# ============================================================
# 13. STAGE 7 - MODEL READY
# ============================================================

model_ready_df = selected_df.copy()


scaler = StandardScaler()


model_ready_df[
    selected_features
] = scaler.fit_transform(
    model_ready_df[
        selected_features
    ]
)


# ============================================================
# 14. PIPELINE PROFILING
# ============================================================

pipeline_stages = [

    (
        "01_Raw_Data",
        raw_df
    ),

    (
        "02_Deduplication",
        clean_df
    ),

    (
        "03_Missing_Value_Imputation",
        imputed_df
    ),

    (
        "04_Outlier_Filtering",
        outlier_df
    ),

    (
        "05_Feature_Engineering",
        feature_df
    ),

    (
        "06_Feature_Selection",
        selected_df
    ),

    (
        "07_Model_Ready",
        model_ready_df
    )

]


pipeline_analysis_df = pd.DataFrame(

    [
        profile_dataset(
            df,
            stage
        )

        for stage, df
        in pipeline_stages
    ]

)


# ============================================================
# 15. INFORMATION RETENTION
# ============================================================

initial_rows = (
    pipeline_analysis_df
    .iloc[0]["Rows"]
)


initial_columns = (
    pipeline_analysis_df
    .iloc[0]["Columns"]
)


initial_cells = (
    pipeline_analysis_df
    .iloc[0]["Total_Cells"]
)


initial_entropy = (
    pipeline_analysis_df
    .iloc[0]["Mean_Entropy"]
)


pipeline_analysis_df[
    "Row_Retention_%"
] = (

    pipeline_analysis_df["Rows"]

    /

    initial_rows

    * 100

)


pipeline_analysis_df[
    "Column_Retention_%"
] = (

    pipeline_analysis_df["Columns"]

    /

    initial_columns

    * 100

)


pipeline_analysis_df[
    "Cell_Retention_%"
] = (

    pipeline_analysis_df["Total_Cells"]

    /

    initial_cells

    * 100

)


pipeline_analysis_df[
    "Information_Retention_%"
] = (

    pipeline_analysis_df[
        "Mean_Entropy"
    ]

    /

    max(
        initial_entropy,
        1e-12
    )

    * 100

)


pipeline_analysis_df[
    "Information_Loss_%"
] = (

    100

    -

    pipeline_analysis_df[
        "Information_Retention_%"
    ]

)


pipeline_analysis_df[
    "Transformation_Loss_%"
] = (

    pipeline_analysis_df[
        "Information_Retention_%"
    ]
    .shift(1)

    -

    pipeline_analysis_df[
        "Information_Retention_%"
    ]

).fillna(0)


# ============================================================
# 16. FEATURE EVOLUTION
# ============================================================

tracked_features = [

    column

    for column in numeric_columns

    if column in feature_df.columns

    and column != "customer_id"

]


feature_evolution_records = []


feature_stages = [

    ("Raw", raw_df),

    ("Cleaned", clean_df),

    ("Imputed", imputed_df),

    ("Outlier_Filtered", outlier_df),

    ("Feature_Engineered", feature_df),

    ("Selected", selected_df)

]


for feature in tracked_features:

    for stage_name, stage_df in feature_stages:

        if feature not in stage_df.columns:
            continue

        series = (
            stage_df[feature]
            .dropna()
        )

        feature_evolution_records.append({

            "Feature":
                feature,

            "Stage":
                stage_name,

            "Mean":
                round(
                    series.mean(),
                    4
                ),

            "Std":
                round(
                    series.std(),
                    4
                ),

            "Variance":
                round(
                    series.var(),
                    4
                ),

            "Unique_Values":
                int(
                    series.nunique()
                ),

            "Missing_Count":
                int(
                    stage_df[
                        feature
                    ].isna().sum()
                ),

            "Entropy":
                round(
                    calculate_entropy(
                        stage_df[
                            feature
                        ]
                    ),
                    5
                )

        })


feature_evolution_df = (
    pd.DataFrame(
        feature_evolution_records
    )
)


# ============================================================
# 17. DISTRIBUTION SHIFT
# ============================================================

distribution_records = []


comparison_pairs = [

    (
        "Raw_to_Cleaned",
        raw_df,
        clean_df
    ),

    (
        "Cleaned_to_Imputed",
        clean_df,
        imputed_df
    ),

    (
        "Imputed_to_Outlier_Filtered",
        imputed_df,
        outlier_df
    )

]


for transition, before, after in comparison_pairs:

    common_columns = sorted(

        set(
            before
            .select_dtypes(
                include=np.number
            )
            .columns
        )

        &

        set(
            after
            .select_dtypes(
                include=np.number
            )
            .columns
        )

    )


    for feature in common_columns:

        a = (
            before[feature]
            .dropna()
        )

        b = (
            after[feature]
            .dropna()
        )


        if (
            len(a) > 5
            and
            len(b) > 5
        ):

            statistic, p_value = (
                ks_2samp(
                    a,
                    b
                )
            )


            distribution_records.append({

                "Transition":
                    transition,

                "Feature":
                    feature,

                "KS_Statistic":
                    round(
                        statistic,
                        5
                    ),

                "P_Value":
                    round(
                        p_value,
                        7
                    ),

                "Distribution_Changed":
                    (
                        "YES"
                        if p_value < 0.05
                        else "NO"
                    )

            })


distribution_shift_df = (
    pd.DataFrame(
        distribution_records
    )
)


# ============================================================
# 18. FEATURE INFORMATION LOSS
# ============================================================

feature_loss_records = []


for feature in tracked_features:

    raw_entropy = (
        calculate_entropy(
            raw_df[feature]
        )
    )


    final_entropy = (
        calculate_entropy(
            feature_df[feature]
        )
    )


    retention = (
        final_entropy
        /
        max(
            raw_entropy,
            1e-12
        )
        *
        100
    )


    feature_loss_records.append({

        "Feature":
            feature,

        "Raw_Entropy":
            round(
                raw_entropy,
                5
            ),

        "Final_Entropy":
            round(
                final_entropy,
                5
            ),

        "Information_Retention_%":
            round(
                retention,
                3
            ),

        "Information_Loss_%":
            round(
                max(
                    0,
                    100 - retention
                ),
                3
            )

    })


feature_information_loss_df = (

    pd.DataFrame(
        feature_loss_records
    )

    .sort_values(
        "Information_Loss_%",
        ascending=False
    )

    .reset_index(
        drop=True
    )

)


# ============================================================
# 19. MUTUAL INFORMATION
# ============================================================

X_mi = (
    feature_df[
        selected_features
    ]
    .copy()
)


y_mi = (
    feature_df["churn"]
    .copy()
)


mi_scores = mutual_info_classif(
    X_mi,
    y_mi,
    random_state=42
)


feature_information_df = (

    pd.DataFrame({

        "Feature":
            selected_features,

        "Mutual_Information":
            mi_scores

    })

    .sort_values(
        "Mutual_Information",
        ascending=False
    )

    .reset_index(
        drop=True
    )

)


feature_information_df[
    "Mutual_Information"
] = (
    feature_information_df[
        "Mutual_Information"
    ]
    .round(5)
)


# ============================================================
# 20. RANDOM FOREST MODEL
# ============================================================

X = (
    feature_df[
        selected_features
    ]
    .copy()
)


y = (
    feature_df["churn"]
    .copy()
)


X_train, X_test, y_train, y_test = (

    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=42,

        stratify=y

    )

)


model = RandomForestClassifier(

    n_estimators=250,

    max_depth=8,

    min_samples_leaf=4,

    random_state=42,

    n_jobs=-1

)


model.fit(
    X_train,
    y_train
)


predictions = (
    model.predict(
        X_test
    )
)


probabilities = (
    model.predict_proba(
        X_test
    )[:, 1]
)


model_metrics_df = pd.DataFrame([{

    "Accuracy":
        accuracy_score(
            y_test,
            predictions
        ),

    "Precision":
        precision_score(
            y_test,
            predictions,
            zero_division=0
        ),

    "Recall":
        recall_score(
            y_test,
            predictions,
            zero_division=0
        ),

    "F1":
        f1_score(
            y_test,
            predictions,
            zero_division=0
        ),

    "ROC_AUC":
        roc_auc_score(
            y_test,
            probabilities
        )

}])


model_metrics_df = (
    model_metrics_df
    .round(4)
)


# ============================================================
# 21. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

feature_importance_df = (

    pd.DataFrame({

        "Feature":
            selected_features,

        "Random_Forest_Importance":
            model.feature_importances_

    })

    .sort_values(
        "Random_Forest_Importance",
        ascending=False
    )

    .reset_index(
        drop=True
    )

)


feature_importance_df[
    "Random_Forest_Importance"
] = (
    feature_importance_df[
        "Random_Forest_Importance"
    ]
    .round(5)
)


# ============================================================
# 22. FEATURE ABLATION
# ============================================================

baseline_f1 = float(
    model_metrics_df.loc[
        0,
        "F1"
    ]
)


ablation_records = []


for feature in selected_features:

    remaining_features = [

        f

        for f in selected_features

        if f != feature

    ]


    ablation_model = RandomForestClassifier(

        n_estimators=120,

        max_depth=8,

        min_samples_leaf=4,

        random_state=42,

        n_jobs=-1

    )


    ablation_model.fit(

        X_train[
            remaining_features
        ],

        y_train

    )


    ablation_predictions = (
        ablation_model.predict(
            X_test[
                remaining_features
            ]
        )
    )


    ablation_f1 = f1_score(

        y_test,

        ablation_predictions,

        zero_division=0

    )


    ablation_records.append({

        "Removed_Feature":
            feature,

        "Baseline_F1":
            round(
                baseline_f1,
                4
            ),

        "Ablated_F1":
            round(
                ablation_f1,
                4
            ),

        "F1_Change":
            round(
                ablation_f1 -
                baseline_f1,
                4
            ),

        "Performance_Drop":
            round(
                baseline_f1 -
                ablation_f1,
                4
            )

    })


ablation_df = (

    pd.DataFrame(
        ablation_records
    )

    .sort_values(
        "Performance_Drop",
        ascending=False
    )

    .reset_index(
        drop=True
    )

)


# ============================================================
# 23. DATA QUALITY SCORE
# ============================================================

quality_records = []


for _, row in (
    pipeline_analysis_df.iterrows()
):

    missing_score = max(

        0,

        100 -
        row["Missing_Rate_%"] *
        5

    )


    duplicate_rate = (

        row["Duplicate_Rows"]

        /

        max(
            row["Rows"],
            1
        )

        *

        100

    )


    duplicate_score = max(

        0,

        100 -
        duplicate_rate *
        5

    )


    information_score = min(

        max(
            row[
                "Information_Retention_%"
            ],
            0
        ),

        100

    )


    overall_quality = (

        missing_score

        +

        duplicate_score

        +

        information_score

    ) / 3


    quality_records.append({

        "Stage":
            row["Stage"],

        "Missing_Data_Score":
            round(
                missing_score,
                2
            ),

        "Duplicate_Score":
            round(
                duplicate_score,
                2
            ),

        "Information_Score":
            round(
                information_score,
                2
            ),

        "Overall_Data_Quality":
            round(
                overall_quality,
                2
            )

    })


data_quality_df = (
    pd.DataFrame(
        quality_records
    )
)


# ============================================================
# 24. FINAL SUMMARY
# ============================================================

largest_loss_row = (
    pipeline_analysis_df.loc[
        pipeline_analysis_df[
            "Transformation_Loss_%"
        ].idxmax()
    ]
)


highest_loss_feature = (
    feature_information_loss_df
    .iloc[0]
)


top_mi_feature = (
    feature_information_df
    .iloc[0]
)


top_model_feature = (
    feature_importance_df
    .iloc[0]
)


final_summary_df = pd.DataFrame([{

    "Initial_Rows":
        int(initial_rows),

    "Final_Rows":
        int(
            len(model_ready_df)
        ),

    "Rows_Removed":
        int(
            initial_rows -
            len(model_ready_df)
        ),

    "Initial_Columns":
        int(initial_columns),

    "Final_Columns":
        int(
            model_ready_df.shape[1]
        ),

    "Total_Information_Loss_%":
        round(
            float(
                pipeline_analysis_df.iloc[-1][
                    "Information_Loss_%"
                ]
            ),
            3
        ),

    "Largest_Transformation_Stage":
        largest_loss_row[
            "Stage"
        ],

    "Largest_Transformation_Loss_%":
        round(
            float(
                largest_loss_row[
                    "Transformation_Loss_%"
                ]
            ),
            3
        ),

    "Highest_Loss_Feature":
        highest_loss_feature[
            "Feature"
        ],

    "Highest_Loss_Feature_%":
        float(
            highest_loss_feature[
                "Information_Loss_%"
            ]
        ),

    "Top_MI_Feature":
        top_mi_feature[
            "Feature"
        ],

    "Top_MI_Value":
        float(
            top_mi_feature[
                "Mutual_Information"
            ]
        ),

    "Top_Model_Feature":
        top_model_feature[
            "Feature"
        ],

    "Top_Model_Importance":
        float(
            top_model_feature[
                "Random_Forest_Importance"
            ]
        ),

    "Model_F1":
        float(
            model_metrics_df.loc[
                0,
                "F1"
            ]
        ),

    "Model_ROC_AUC":
        float(
            model_metrics_df.loc[
                0,
                "ROC_AUC"
            ]
        )

}])


# ============================================================
# 25. CORRELATION EVOLUTION
# ============================================================

correlation_records = []


for stage_name, stage_df in [

    ("Raw", raw_df),

    ("Cleaned", clean_df),

    ("Imputed", imputed_df),

    ("Outlier_Filtered", outlier_df),

    ("Feature_Engineered", feature_df),

    ("Selected", selected_df)

]:

    correlation_matrix = (
        stage_df
        .select_dtypes(
            include=np.number
        )
        .corr()
    )


    columns = (
        correlation_matrix
        .columns
        .tolist()
    )


    for i in range(
        len(columns)
    ):

        for j in range(
            i + 1,
            len(columns)
        ):

            correlation_records.append({

                "Stage":
                    stage_name,

                "Feature_A":
                    columns[i],

                "Feature_B":
                    columns[j],

                "Correlation":
                    round(
                        float(
                            correlation_matrix.iloc[
                                i,
                                j
                            ]
                        ),
                        5
                    )

            })


correlation_evolution_df = (
    pd.DataFrame(
        correlation_records
    )
)


# ============================================================
# 26. DISPLAY ANALYSIS DATAFRAMES
# ============================================================

print("\n" + "=" * 100)
print("PIPELINE ANALYSIS")
print("=" * 100)

display(
    pipeline_analysis_df.round(3)
)


print("\n" + "=" * 100)
print("FEATURE EVOLUTION")
print("=" * 100)

display(
    feature_evolution_df.head(30)
)


print("\n" + "=" * 100)
print("DISTRIBUTION SHIFT")
print("=" * 100)

display(
    distribution_shift_df.head(30)
)


print("\n" + "=" * 100)
print("FEATURE INFORMATION LOSS")
print("=" * 100)

display(
    feature_information_loss_df
)


print("\n" + "=" * 100)
print("MUTUAL INFORMATION")
print("=" * 100)

display(
    feature_information_df
)


print("\n" + "=" * 100)
print("MODEL METRICS")
print("=" * 100)

display(
    model_metrics_df
)


print("\n" + "=" * 100)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 100)

display(
    feature_importance_df
)


print("\n" + "=" * 100)
print("FEATURE ABLATION")
print("=" * 100)

display(
    ablation_df
)


print("\n" + "=" * 100)
print("DATA QUALITY")
print("=" * 100)

display(
    data_quality_df
)


print("\n" + "=" * 100)
print("FINAL SUMMARY")
print("=" * 100)

display(
    final_summary_df
)


# ============================================================
# 27. VISUALIZATION 1
# INFORMATION RETENTION
# ============================================================

plt.figure(
    figsize=(13, 6)
)

plt.plot(

    pipeline_analysis_df["Stage"],

    pipeline_analysis_df[
        "Information_Retention_%"
    ],

    marker="o",

    linewidth=2

)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylabel(
    "Information Retention (%)"
)

plt.xlabel(
    "Pipeline Stage"
)

plt.title(
    "Information Retention Across the ML Pipeline"
)

plt.ylim(
    0,
    110
)

plt.grid(
    alpha=0.25
)

save_plot(
    "01_information_retention.png"
)


# ============================================================
# 28. VISUALIZATION 2
# TRANSFORMATION LOSS
# ============================================================

plt.figure(
    figsize=(13, 6)
)

plt.bar(

    pipeline_analysis_df["Stage"],

    pipeline_analysis_df[
        "Transformation_Loss_%"
    ]

)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylabel(
    "Transformation Loss (%)"
)

plt.title(
    "Information Loss Introduced by Each Transformation"
)

save_plot(
    "02_transformation_loss.png"
)


# ============================================================
# 29. VISUALIZATION 3
# ROW RETENTION
# ============================================================

plt.figure(
    figsize=(13, 6)
)

plt.plot(

    pipeline_analysis_df["Stage"],

    pipeline_analysis_df[
        "Row_Retention_%"
    ],

    marker="o",

    linewidth=2

)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylabel(
    "Rows Retained (%)"
)

plt.title(
    "Dataset Row Retention Across Pipeline"
)

plt.ylim(
    0,
    105
)

plt.grid(
    alpha=0.25
)

save_plot(
    "03_row_retention.png"
)


# ============================================================
# 30. VISUALIZATION 4
# DATA QUALITY
# ============================================================

plt.figure(
    figsize=(13, 6)
)

plt.plot(

    data_quality_df["Stage"],

    data_quality_df[
        "Overall_Data_Quality"
    ],

    marker="o",

    linewidth=2

)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylabel(
    "Data Quality Score"
)

plt.title(
    "Data Quality Evolution"
)

plt.ylim(
    0,
    105
)

plt.grid(
    alpha=0.25
)

save_plot(
    "04_data_quality.png"
)


# ============================================================
# 31. VISUALIZATION 5
# FEATURE INFORMATION LOSS
# ============================================================

plot_loss = (
    feature_information_loss_df
    .sort_values(
        "Information_Loss_%"
    )
)


plt.figure(
    figsize=(10, 7)
)

plt.barh(

    plot_loss["Feature"],

    plot_loss[
        "Information_Loss_%"
    ]

)

plt.xlabel(
    "Information Loss (%)"
)

plt.title(
    "Information Loss by Feature"
)

save_plot(
    "05_feature_information_loss.png"
)


# ============================================================
# 32. VISUALIZATION 6
# MUTUAL INFORMATION
# ============================================================

plot_mi = (
    feature_information_df
    .sort_values(
        "Mutual_Information"
    )
)


plt.figure(
    figsize=(10, 7)
)

plt.barh(

    plot_mi["Feature"],

    plot_mi[
        "Mutual_Information"
    ]

)

plt.xlabel(
    "Mutual Information"
)

plt.title(
    "Feature Information Value"
)

save_plot(
    "06_mutual_information.png"
)


# ============================================================
# 33. VISUALIZATION 7
# MISSING VALUES
# ============================================================

missing_values = (
    raw_df
    .isna()
    .sum()
)


missing_values = (
    missing_values[
        missing_values > 0
    ]
    .sort_values()
)


plt.figure(
    figsize=(10, 6)
)

plt.barh(

    missing_values.index,

    missing_values.values

)

plt.xlabel(
    "Missing Values"
)

plt.title(
    "Raw Dataset Missing-Value Profile"
)

save_plot(
    "07_missing_values.png"
)


# ============================================================
# 34. VISUALIZATION 8
# CORRELATION MATRIX
# ============================================================

correlation_matrix = (
    feature_df[
        selected_features +
        ["churn"]
    ]
    .corr()
)


plt.figure(
    figsize=(11, 9)
)

plt.imshow(
    correlation_matrix,
    aspect="auto"
)

plt.colorbar(
    label="Correlation"
)

plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=45,

    ha="right"

)

plt.yticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns

)

plt.title(
    "Feature Correlation Structure"
)

save_plot(
    "08_correlation_matrix.png"
)


# ============================================================
# 35. VISUALIZATION 9
# DISTRIBUTION SHIFT
# ============================================================

if not distribution_shift_df.empty:

    shift_summary = (

        distribution_shift_df

        .assign(
            Changed=
            distribution_shift_df[
                "Distribution_Changed"
            ].eq("YES")
        )

        .groupby(
            "Transition"
        )["Changed"]

        .sum()

        .reset_index(
            name="Changed_Features"
        )

    )


    plt.figure(
        figsize=(10, 6)
    )


    plt.bar(

        shift_summary[
            "Transition"
        ],

        shift_summary[
            "Changed_Features"
        ]

    )


    plt.xticks(
        rotation=30,
        ha="right"
    )


    plt.ylabel(
        "Features With Distribution Change"
    )


    plt.title(
        "Distribution Shifts Detected by KS Test"
    )


    save_plot(
        "09_distribution_shift.png"
    )


# ============================================================
# 36. VISUALIZATION 10
# FEATURE ABLATION
# ============================================================

plot_ablation = (
    ablation_df
    .sort_values(
        "Performance_Drop"
    )
)


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    plot_ablation[
        "Removed_Feature"
    ],

    plot_ablation[
        "Performance_Drop"
    ]

)


plt.xlabel(
    "F1 Performance Drop"
)


plt.title(
    "Feature Ablation Impact on Model Performance"
)


save_plot(
    "10_feature_ablation.png"
)


# ============================================================
# 37. EXPORT ALL DATAFRAMES
# ============================================================

dataframes = {

    "raw_dataset":
        raw_df,

    "clean_dataset":
        clean_df,

    "imputed_dataset":
        imputed_df,

    "outlier_filtered_dataset":
        outlier_df,

    "feature_engineered_dataset":
        feature_df,

    "selected_dataset":
        selected_df,

    "model_ready_dataset":
        model_ready_df,

    "pipeline_analysis":
        pipeline_analysis_df,

    "feature_evolution":
        feature_evolution_df,

    "distribution_shift":
        distribution_shift_df,

    "correlation_evolution":
        correlation_evolution_df,

    "feature_information_loss":
        feature_information_loss_df,

    "feature_information":
        feature_information_df,

    "feature_importance":
        feature_importance_df,

    "ablation_analysis":
        ablation_df,

    "data_quality":
        data_quality_df,

    "model_metrics":
        model_metrics_df,

    "final_summary":
        final_summary_df

}


for name, dataframe in dataframes.items():

    dataframe.to_csv(

        DATA_DIR /
        f"{name}.csv",

        index=False

    )


# ============================================================
# 38. EXCEL REPORT
# ============================================================

excel_path = (

    REPORT_DIR /
    "information_loss_observatory.xlsx"

)


with pd.ExcelWriter(
    excel_path,
    engine="openpyxl"
) as writer:

    for name, dataframe in dataframes.items():

        dataframe.to_excel(

            writer,

            sheet_name=name[:31],

            index=False

        )


# ============================================================
# 39. TEXT REPORT
# ============================================================

summary = final_summary_df.iloc[0]


report_text = f"""

INFORMATION LOSS OBSERVATORY
============================

Data Transformation & ML Pipeline Intelligence


DATASET
-------

Initial Rows:
{summary["Initial_Rows"]}

Final Rows:
{summary["Final_Rows"]}

Rows Removed:
{summary["Rows_Removed"]}

Initial Columns:
{summary["Initial_Columns"]}

Final Columns:
{summary["Final_Columns"]}


INFORMATION ANALYSIS
--------------------

Measured Information Loss:
{summary["Total_Information_Loss_%"]}%

Largest Transformation Stage:
{summary["Largest_Transformation_Stage"]}

Largest Transformation Loss:
{summary["Largest_Transformation_Loss_%"]}%


FEATURE ANALYSIS
----------------

Highest Loss Feature:
{summary["Highest_Loss_Feature"]}

Highest Feature Loss:
{summary["Highest_Loss_Feature_%"]}%


Top Mutual Information Feature:
{summary["Top_MI_Feature"]}

Mutual Information:
{summary["Top_MI_Value"]}


Top Random Forest Feature:
{summary["Top_Model_Feature"]}

Random Forest Importance:
{summary["Top_Model_Importance"]}


MODEL PERFORMANCE
-----------------

Accuracy:
{model_metrics_df.loc[0, "Accuracy"]}

Precision:
{model_metrics_df.loc[0, "Precision"]}

Recall:
{model_metrics_df.loc[0, "Recall"]}

F1:
{model_metrics_df.loc[0, "F1"]}

ROC-AUC:
{model_metrics_df.loc[0, "ROC_AUC"]}


METHODOLOGY NOTE
----------------

Information retention/loss is a project-defined analytical proxy
based on the mean entropy of numerical feature distributions.

It should not be interpreted as a universal measurement of
information-theoretic loss.

The dataset used by this demonstration is synthetic.

The model results therefore demonstrate pipeline analysis and
engineering methodology rather than real-world predictive
performance.
"""


with open(

    REPORT_DIR /
    "analysis_report.txt",

    "w",

    encoding="utf-8"

) as file:

    file.write(
        report_text
    )


# ============================================================
# 40. FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 100)
print("INFORMATION LOSS OBSERVATORY COMPLETED SUCCESSFULLY")
print("=" * 100)

print(
    f"Initial Dataset       : {raw_df.shape}"
)

print(
    f"Final Dataset         : {model_ready_df.shape}"
)

print(
    f"Information Loss      : {summary['Total_Information_Loss_%']}%"
)

print(
    f"Largest Loss Stage    : {summary['Largest_Transformation_Stage']}"
)

print(
    f"Highest Loss Feature  : {summary['Highest_Loss_Feature']}"
)

print(
    f"Top MI Feature        : {summary['Top_MI_Feature']}"
)

print(
    f"Top Model Feature     : {summary['Top_Model_Feature']}"
)

print(
    f"Model F1              : {summary['Model_F1']}"
)

print(
    f"Model ROC-AUC         : {summary['Model_ROC_AUC']}"
)

print()
print(
    f"DATA DIRECTORY        : {DATA_DIR}"
)

print(
    f"REPORT DIRECTORY      : {REPORT_DIR}"
)

print(
    f"EXCEL REPORT          : {excel_path}"
)

print(
    f"FIGURES DIRECTORY     : {FIGURE_DIR}"
)

print("=" * 100)
print("ALL ANALYSIS COMPLETED")
print("=" * 100)
```
